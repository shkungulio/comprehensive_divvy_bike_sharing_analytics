## Cell 1 — Imports, Configuration, and Helpers

**In plain English:** This cell sets up the "control panel" for the entire analysis before any real work happens. It connects Python to the PostgreSQL database, sets a consistent visual style for every chart that follows, creates a folder where results will be saved, and — importantly — builds a single, adjustable data-quality filter (`WHERE` clause) that every query in the notebook will reuse. Two small reusable functions are defined: one that safely runs a SQL query and always closes the database connection (even if something fails), and one that saves any chart to disk with print confirmation.

**Why it matters for the business:** Centralizing the "what counts as a valid trip" rule in one place (instead of repeating it in every query) means a reviewer or teammate can change the analysis population — e.g., include flagged anomalies, or cap ride duration — by editing one line, and every downstream number and chart updates consistently. That is exactly the kind of change a stakeholder asks for in a real review meeting.

**Skills demonstrated:**
- Database connectivity from Python (`psycopg2`) with proper connection lifecycle management (`try/except/finally`)
- Configuration-driven, reproducible analysis design (adjustable filters and constants set in one place)
- Defensive/production-style coding (custom exception handling, guarding against undefined variables)
- Function design for code reuse (DRY principle)
- Environment/output management (`os.makedirs`, file path handling)
- Data visualization setup and styling conventions (`seaborn` themes)


```python
import os
import warnings
from datetime import datetime, timezone
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import psycopg2
import seaborn as sns
from IPython.display import Markdown, display
import config

warnings.filterwarnings("ignore", category=UserWarning)
sns.set_theme(style="whitegrid", context="notebook")
PROJECT_FOLDER = os.path.abspath(os.path.join(os.getcwd(), os.pardir))
REPORTS_FOLDER = os.path.join(PROJECT_FOLDER, "reports", "exploratory_analysis_reports")
os.makedirs(REPORTS_FOLDER, exist_ok=True)
SCHEMA = "divvy"
EXCLUDE_FLAGGED_ANOMALIES = True
MIN_DURATION_MINUTES = None
MAX_DURATION_MINUTES = None
TOP_N_STATIONS = 15
RUN_TIMESTAMP = datetime.now(timezone.utc).isoformat(timespec="seconds")

def run_query(sql, params=None):
    try:
        conn = psycopg2.connect(**config.config_divvy())
        return pd.read_sql_query(sql, conn, params=params)
    except Exception as error:
        raise ConnectionError(f"Could not query divvy_db: {error}") from error
    finally:
        if "conn" in locals():
            conn.close()

def savefig(filename):
    path = os.path.join(REPORTS_FOLDER, filename)
    plt.savefig(path, dpi=160, bbox_inches="tight")
    print("Saved:", path)

filters = ["f.duration_minutes > 0"]
if EXCLUDE_FLAGGED_ANOMALIES: filters.append("COALESCE(f.is_anomalous, FALSE) = FALSE")
if MIN_DURATION_MINUTES is not None: filters.append(f"f.duration_minutes >= {float(MIN_DURATION_MINUTES)}")
if MAX_DURATION_MINUTES is not None: filters.append(f"f.duration_minutes <= {float(MAX_DURATION_MINUTES)}")
WHERE = " AND ".join(filters)
print("Reports:", REPORTS_FOLDER, "\nPopulation:", WHERE)
```

## Cell 2 — Coverage, Warehouse Views, and Member vs. Casual Behavior

**In plain English:** This cell first runs a "data health check" — how many trips are in the warehouse overall, what date range they cover, and what percentage were flagged as bad data or are missing a station. It then previews a pre-built summary view of daily ridership. Finally, it runs the notebook's core comparison: annual members vs. casual (pay-per-ride) riders, measuring how many rides each group takes, how long their typical ride is (using both average and median, plus the 95th-percentile "long ride" cutoff), what share of their riding happens on weekends, and how often they make round trips. Two bar charts visualize ride volume and typical duration side by side.

**Why it matters for the business:** This single cell answers the question every bike-share operator asks first: "Are members and casual riders using the system differently?" Comparing average duration against median duration is a subtle but important analytical choice — a large gap between the two signals a "long tail" of unusually long rides pulling the average up, which changes how you'd interpret pricing or fleet-rebalancing decisions.

**Skills demonstrated:**
- Data-quality auditing via SQL (row counts, date range checks, percentage of missing/flagged values, all computed inline with `AVG(boolean::int)`)
- Multi-table `JOIN`s across a star-schema warehouse (fact table + two dimension tables)
- Statistical aggregation in SQL, including percentiles (`PERCENTILE_CONT`) for distribution analysis beyond a simple average
- Querying and validating pre-built database views (a sign of working within an existing data model rather than only writing ad-hoc queries)
- Pandas post-processing of query results (computing a derived "share of total" metric)
- Comparative, grouped visualization with `seaborn`/`matplotlib`, including a two-panel figure layout


```python
coverage = run_query(f'''
SELECT COUNT(*) total_trips, MIN(started_at) first_start, MAX(started_at) last_start,
 COUNT(DISTINCT start_date_key) active_dates, COUNT(DISTINCT member_type_key) member_segments,
 COUNT(DISTINCT ride_type_key) bike_types, AVG((is_anomalous IS TRUE)::int)*100 flagged_anomaly_pct,
 AVG((start_station_key IS NULL OR end_station_key IS NULL)::int)*100 missing_station_pct
FROM {SCHEMA}.fact_trip;
''')
display(coverage)
display(run_query(f"SELECT full_date, member_casual, ride_count, avg_duration_minutes FROM {SCHEMA}.vw_daily_rides ORDER BY full_date DESC, member_casual LIMIT 10"))

segments = run_query(f'''
SELECT mt.member_casual, COUNT(*) rides, AVG(f.duration_minutes) avg_duration_min,
 PERCENTILE_CONT(.5) WITHIN GROUP (ORDER BY f.duration_minutes) median_duration_min,
 PERCENTILE_CONT(.95) WITHIN GROUP (ORDER BY f.duration_minutes) p95_duration_min,
 AVG(d.is_weekend::int)*100 weekend_ride_pct, AVG(COALESCE(f.is_round_trip,FALSE)::int)*100 round_trip_pct
FROM {SCHEMA}.fact_trip f JOIN {SCHEMA}.dim_member_type mt ON f.member_type_key=mt.member_type_key
JOIN {SCHEMA}.dim_date d ON f.start_date_key=d.date_key WHERE {WHERE}
GROUP BY mt.member_casual ORDER BY rides DESC;
''')
segments["ride_share_pct"] = 100*segments.rides/segments.rides.sum()
display(segments.round(2))
fig,axes=plt.subplots(1,2,figsize=(13,4))
sns.barplot(data=segments,x="member_casual",y="rides",hue="member_casual",legend=False,palette="Set2",ax=axes[0])
axes[0].set(title="Ride Volume by Rider Type",xlabel="",ylabel="Rides"); axes[0].ticklabel_format(style="plain",axis="y")
plot=segments.melt(id_vars="member_casual",value_vars=["avg_duration_min","median_duration_min"])
sns.barplot(data=plot,x="member_casual",y="value",hue="variable",palette="Set2",ax=axes[1])
axes[1].set(title="Typical Ride Duration",xlabel="",ylabel="Minutes"); axes[1].legend(title="Metric",labels=["Average","Median"])
plt.tight_layout(); savefig("01_member_casual.png"); plt.show()
```

## Cell 3 — Timing, Seasonality, and Bike Types

**In plain English:** This cell answers "when do people ride, and on what?" It pulls one detailed dataset from the warehouse (rides broken down by exact date, day of week, hour, month, rider type, and bike type) and then reshapes that single result four different ways in Python — by day of week, by hour of day, by month (to see seasonal trends across the year), and by bike type (classic bike, electric bike, scooter, etc.). All six resulting charts split members from casual riders, so patterns like "members peak during weekday commute hours" versus "casual riders peak on weekend afternoons" become visible at a glance.

**Why it matters for the business:** This is the kind of chart set that goes straight into an operations or marketing deck — it tells the business when to schedule bike rebalancing crews, when to run promotions, and which bike types to prioritize purchasing, broken down by the two customer segments that behave very differently.

**Skills demonstrated:**
- Efficient query design: pulling one appropriately-grained dataset from the database rather than running six separate expensive queries
- Three-way table `JOIN`s in a dimensional data warehouse
- Pandas `groupby().agg()` for flexible, multi-dimensional re-aggregation of a single dataset (a core data-manipulation skill)
- Handling ordered categorical data correctly (forcing weekday order instead of alphabetical) and reconstructing proper dates from separate year/month columns
- Building multi-panel dashboards (a 2×3 grid of six related charts in one figure) with `matplotlib`/`seaborn`
- Time-series and seasonality analysis, and segmented ("faceted") comparison across two customer groups simultaneously


```python
profiles = run_query(f'''
SELECT d.full_date,d.day_of_week,d.day_name,d.year,d.month,d.month_name,f.start_hour,mt.member_casual,rt.rideable_type,
 COUNT(*) rides,AVG(f.duration_minutes) avg_duration_min
FROM {SCHEMA}.fact_trip f JOIN {SCHEMA}.dim_date d ON f.start_date_key=d.date_key
JOIN {SCHEMA}.dim_member_type mt ON f.member_type_key=mt.member_type_key
JOIN {SCHEMA}.dim_ride_type rt ON f.ride_type_key=rt.ride_type_key WHERE {WHERE}
GROUP BY d.full_date,d.day_of_week,d.day_name,d.year,d.month,d.month_name,f.start_hour,mt.member_casual,rt.rideable_type;
''')
order=["Monday","Tuesday","Wednesday","Thursday","Friday","Saturday","Sunday"]
daily=profiles.groupby(["day_of_week","day_name","member_casual"],as_index=False).agg(rides=("rides","sum"),duration=("avg_duration_min","mean"))
daily["day_name"]=pd.Categorical(daily.day_name,categories=order,ordered=True)
hourly=profiles.groupby(["start_hour","member_casual"],as_index=False).agg(rides=("rides","sum"),duration=("avg_duration_min","mean"))
monthly=profiles.groupby(["year","month","member_casual"],as_index=False).agg(rides=("rides","sum"),duration=("avg_duration_min","mean"))
monthly["date"]=pd.to_datetime(dict(year=monthly.year,month=monthly.month,day=1))
bike=profiles.groupby(["rideable_type","member_casual"],as_index=False).agg(rides=("rides","sum"),duration=("avg_duration_min","mean"))

fig,axes=plt.subplots(2,3,figsize=(18,9))
sns.lineplot(data=daily.sort_values("day_of_week"),x="day_name",y="rides",hue="member_casual",marker="o",ax=axes[0,0]); axes[0,0].set(title="Volume by Day",xlabel="",ylabel="Rides"); axes[0,0].tick_params(axis="x",rotation=30)
sns.lineplot(data=daily.sort_values("day_of_week"),x="day_name",y="duration",hue="member_casual",marker="o",ax=axes[0,1]); axes[0,1].set(title="Duration by Day",xlabel="",ylabel="Minutes"); axes[0,1].tick_params(axis="x",rotation=30)
sns.lineplot(data=hourly,x="start_hour",y="rides",hue="member_casual",marker="o",ax=axes[0,2]); axes[0,2].set(title="Volume by Hour",xlabel="Start hour",ylabel="Rides",xticks=range(0,24,2))
sns.lineplot(data=hourly,x="start_hour",y="duration",hue="member_casual",marker="o",ax=axes[1,0]); axes[1,0].set(title="Duration by Hour",xlabel="Start hour",ylabel="Minutes",xticks=range(0,24,2))
sns.lineplot(data=monthly,x="date",y="rides",hue="member_casual",marker="o",ax=axes[1,1]); axes[1,1].set(title="Monthly Volume",xlabel="",ylabel="Rides"); axes[1,1].tick_params(axis="x",rotation=30)
sns.barplot(data=bike,x="rideable_type",y="rides",hue="member_casual",ax=axes[1,2]); axes[1,2].set(title="Bike-Type Mix",xlabel="Bike type",ylabel="Rides")
plt.tight_layout(); savefig("02_time_and_bike_patterns.png"); plt.show()
display(bike.sort_values(["rideable_type","rides"],ascending=[True,False]))
```

## Cell 4 — Station Activity and Net Flows

**In plain English:** This cell figures out which stations matter most and where bikes tend to pile up or run out. It builds two temporary, named sub-queries ("starts" and "ends") — one counting how many rides began at each station, one counting how many ended there — then combines them per station to compute a "net outflow" (rides started minus rides ended). A station with a big positive number is bleeding bikes out (more people take bikes than return them there); a big negative number means bikes pile up. It visualizes both the busiest stations overall and the stations with the most extreme imbalance in either direction.

**Why it matters for the business:** This is a classic bike-share logistics problem — knowing which stations chronically run empty or overflow tells operations teams exactly where to send rebalancing trucks, which is a direct, dollar-relevant business decision, not just an interesting statistic.

**Skills demonstrated:**
- Advanced SQL using Common Table Expressions (`WITH` clauses) to organize a multi-step calculation readably
- `LEFT JOIN` combined with `COALESCE` to correctly handle stations that only ever appear as a start or an end (avoiding silent data loss from an inner join)
- Deriving a meaningful business metric (net outflow) directly in SQL rather than after the fact
- Pandas `nsmallest`/`nlargest` for surfacing the most extreme cases in a dataset
- Diverging/two-directional bar chart design with conditional coloring (`numpy.where`) to visually separate "bleeding" stations from "absorbing" ones
- Horizontal stacked bar charts for comparing two related quantities (starts and ends) per category


```python
display(run_query(f"SELECT station_id,station_name,full_date,rides_started,rides_ended,net_outflow FROM {SCHEMA}.vw_station_flow ORDER BY full_date DESC,ABS(net_outflow) DESC LIMIT 10"))
station_flow=run_query(f'''
WITH starts AS (SELECT start_station_key station_key,COUNT(*) rides_started FROM {SCHEMA}.fact_trip f WHERE {WHERE} AND start_station_key IS NOT NULL GROUP BY start_station_key),
ends AS (SELECT end_station_key station_key,COUNT(*) rides_ended FROM {SCHEMA}.fact_trip f WHERE {WHERE} AND end_station_key IS NOT NULL GROUP BY end_station_key)
SELECT s.station_id,s.station_name,COALESCE(st.rides_started,0) rides_started,COALESCE(en.rides_ended,0) rides_ended,
 COALESCE(st.rides_started,0)-COALESCE(en.rides_ended,0) net_outflow,COALESCE(st.rides_started,0)+COALESCE(en.rides_ended,0) total_activity
FROM {SCHEMA}.dim_station s LEFT JOIN starts st ON s.station_key=st.station_key LEFT JOIN ends en ON s.station_key=en.station_key
WHERE COALESCE(st.rides_started,0)+COALESCE(en.rides_ended,0)>0 ORDER BY total_activity DESC;
''')
display(station_flow.head(TOP_N_STATIONS))
top=station_flow.head(TOP_N_STATIONS).sort_values("total_activity")
flows=pd.concat([station_flow.nsmallest(TOP_N_STATIONS,"net_outflow"),station_flow.nlargest(TOP_N_STATIONS,"net_outflow")]).drop_duplicates().sort_values("net_outflow")
fig,axes=plt.subplots(1,2,figsize=(18,8))
names=top.station_name.fillna(top.station_id); axes[0].barh(names,top.rides_started,label="Starts"); axes[0].barh(names,top.rides_ended,left=top.rides_started,label="Ends")
axes[0].set(title="Most-Used Stations",xlabel="Rides",ylabel=""); axes[0].legend()
axes[1].barh(flows.station_name.fillna(flows.station_id),flows.net_outflow,color=np.where(flows.net_outflow>=0,"#E45756","#54A24B")); axes[1].axvline(0,color="black",lw=.8)
axes[1].set(title="Largest Net Flows (starts minus ends)",xlabel="Net outflow",ylabel="")
plt.tight_layout(); savefig("03_station_flows.png"); plt.show()
```

## Cell 5 — Seasonal Anomalies and Round-Trip Behavior

**In plain English:** This cell does two things. First, it hunts for unusual days — using each rider segment's own trailing 28-day average and standard deviation as a personalized "normal range," it flags any day whose ride count is at least 3 standard deviations away from that recent normal (a "z-score" of 3+, a standard statistics threshold for "this is genuinely unusual, not just noise"). Those flagged days are plotted as highlighted points on a trend line. Second, it measures round trips — rides that start and end at the same station — broken down by rider type, bike type, and whether it was a weekday or weekend, to see who's more likely to be "just riding around" versus getting somewhere.

**Why it matters for the business:** Automated anomaly detection is far more scalable than a human eyeballing a chart for "weird-looking days" — it's the same statistical technique (rolling z-scores) used in real-world monitoring and alerting systems. The round-trip analysis, meanwhile, distinguishes recreational/leisure usage from transportation usage, which matters for how a company markets to and prices each segment.

**Skills demonstrated:**
- Statistical anomaly/outlier detection using rolling baselines and z-scores — a technique directly transferable to time-series monitoring, fraud detection, and QA/alerting work
- Pandas `groupby().transform()` for computing per-group rolling statistics without collapsing the original row-level data
- SQL conditional aggregation using the `FILTER` clause — a cleaner alternative to `CASE WHEN` for computing multiple conditional metrics in one pass
- Threshold-based flagging and ranking of results by statistical significance
- Layered time-series visualization (raw trend + smoothed baseline + highlighted outliers in one chart)
- Segmented behavioral analysis (rider type × bike type × weekday/weekend) to support customer-segmentation insights


```python
trends=run_query(f'''
SELECT d.full_date,mt.member_casual,COUNT(*) rides,AVG(COALESCE(f.is_anomalous,FALSE)::int)*100 flagged_anomaly_pct
FROM {SCHEMA}.fact_trip f JOIN {SCHEMA}.dim_date d ON f.start_date_key=d.date_key
JOIN {SCHEMA}.dim_member_type mt ON f.member_type_key=mt.member_type_key WHERE f.duration_minutes>0
GROUP BY d.full_date,mt.member_casual ORDER BY d.full_date,mt.member_casual;
''')
trends.full_date=pd.to_datetime(trends.full_date)
trends["baseline"]=trends.groupby("member_casual").rides.transform(lambda x:x.rolling(28,min_periods=14).mean())
trends["std"]=trends.groupby("member_casual").rides.transform(lambda x:x.rolling(28,min_periods=14).std())
trends["z_score"]=(trends.rides-trends.baseline)/trends["std"]
anomalies=trends[trends.z_score.abs()>=3].sort_values("z_score",key=lambda x:x.abs(),ascending=False)
display(anomalies.head(20))
fig,ax=plt.subplots(figsize=(15,5)); sns.lineplot(data=trends,x="full_date",y="rides",hue="member_casual",alpha=.35,ax=ax); sns.lineplot(data=trends,x="full_date",y="baseline",hue="member_casual",legend=False,lw=2.5,ax=ax)
if not anomalies.empty: sns.scatterplot(data=anomalies,x="full_date",y="rides",hue="member_casual",legend=False,s=65,ax=ax)
ax.set(title="Daily Rides and 28-Day Baseline (points: absolute z score at least 3)",xlabel="",ylabel="Rides")
plt.tight_layout(); savefig("04_daily_anomalies.png"); plt.show()

round_trips=run_query(f'''
SELECT mt.member_casual,rt.rideable_type,d.is_weekend,COUNT(*) all_rides,
 COUNT(*) FILTER (WHERE COALESCE(f.is_round_trip,FALSE)) round_trip_rides,
 AVG(f.duration_minutes) FILTER (WHERE COALESCE(f.is_round_trip,FALSE)) round_trip_avg_duration_min,
 100.0*COUNT(*) FILTER (WHERE COALESCE(f.is_round_trip,FALSE))/COUNT(*) round_trip_pct
FROM {SCHEMA}.fact_trip f JOIN {SCHEMA}.dim_member_type mt ON f.member_type_key=mt.member_type_key
JOIN {SCHEMA}.dim_ride_type rt ON f.ride_type_key=rt.ride_type_key JOIN {SCHEMA}.dim_date d ON f.start_date_key=d.date_key
WHERE {WHERE} GROUP BY mt.member_casual,rt.rideable_type,d.is_weekend ORDER BY round_trip_pct DESC;
''')
display(round_trips.round(2))
fig,ax=plt.subplots(figsize=(10,5)); sns.barplot(data=round_trips,x="rideable_type",y="round_trip_pct",hue="member_casual",ax=ax)
ax.set(title="Round-Trip Rate by Bike Type and Rider",xlabel="Bike type",ylabel="Round-trip rides percent")
plt.tight_layout(); savefig("05_round_trip_behavior.png"); plt.show()
```

## Cell 6 — Reproducible Findings Summary

**In plain English:** The last cell automatically writes an executive summary — pulling out the single busiest hour, the single busiest month, the station with the biggest bike surplus/deficit, and the total number of unusual days detected — and renders it as clean, readable bullet points, timestamped so anyone can see exactly when the analysis was run. It then exports every underlying results table to CSV and confirms that all five charts have been saved, so the entire analysis can be handed off, reviewed, or re-run later without re-executing the notebook.

**Why it matters for the business:** This turns a technical notebook into a stakeholder-ready deliverable. A manager or non-technical reviewer doesn't need to read SQL or Python — they get a five-bullet summary plus exportable spreadsheets and images they can drop straight into a slide deck or report.

**Skills demonstrated:**
- Programmatic executive-summary generation (auto-writing plain-language findings from computed results, not hard-coded text)
- Dynamic in-notebook reporting using rendered Markdown output
- Automated, batch file export (looping over a dictionary of DataFrames to CSV) for reproducibility and downstream sharing
- End-to-end analytics pipeline thinking: this notebook produces artifacts (tables + charts + a summary) meant to be consumed outside the notebook itself, reflecting how real analytics deliverables get handed to stakeholders


```python
outflow=station_flow.loc[station_flow.net_outflow.idxmax()]; inflow=station_flow.loc[station_flow.net_outflow.idxmin()]
peak_hour=hourly.groupby("start_hour",as_index=False).rides.sum().nlargest(1,"rides").iloc[0]
peak_month=monthly.groupby("date",as_index=False).rides.sum().nlargest(1,"rides").iloc[0]
facts=[f"Analysis run: {RUN_TIMESTAMP}",f"Valid analysis trips: {int(segments.rides.sum()):,}",f"Peak hour: {int(peak_hour.start_hour):02d}:00",f"Peak month: {peak_month.date:%B %Y}",f"Largest net outflow: {outflow.station_name or outflow.station_id} ({int(outflow.net_outflow):,})",f"Largest net inflow: {inflow.station_name or inflow.station_id} ({int(inflow.net_outflow):,})",f"Statistical anomaly rows: {len(anomalies):,}"]
display(Markdown("### Key outputs\n"+"\n".join("- "+x for x in facts)))
for name,frame in {"member_casual_summary":segments,"bike_type_summary":bike,"station_flow_summary":station_flow,"daily_statistical_anomalies":anomalies}.items(): frame.to_csv(os.path.join(REPORTS_FOLDER,name+".csv"),index=False)
print("Saved tables and five figures to:",REPORTS_FOLDER)
```

## How this notebook reads as a portfolio piece

Taken as a whole, this notebook demonstrates the full analyst workflow that interviewers look for:

1. **Explicit scoping and assumptions** (population filters stated up front) before any numbers are shown.
2. **SQL-first analytics** — pushing aggregation, joins, window functions (`PERCENTILE_CONT`), CTEs, and conditional aggregation (`FILTER`) into the database rather than pulling raw rows into Python, which is the efficient, production-appropriate pattern for large datasets.
3. **Python for what SQL isn't built for** — multi-dimensional reshaping (`groupby`/`transform`), statistical modeling (rolling z-scores), and all visualization.
4. **Segment-aware analysis throughout** — member vs. casual comparisons are threaded through every section rather than treated as a one-off.
5. **Business framing over raw statistics** — every metric (net outflow, round-trip rate, anomaly flags) is tied back to an operational or business question (rebalancing, segmentation, monitoring).
6. **Reproducibility and handoff** — configurable filters, timestamped runs, and automated export of tables/figures/summary, so the work doesn't just live in one person's head or one notebook run.

**Suggested one-line summary for a resume or LinkedIn project post:** *"Built a SQL + Python exploratory analysis pipeline on a PostgreSQL data warehouse of Divvy bike-share trips, combining star-schema queries (joins, CTEs, percentiles, conditional aggregation) with statistical anomaly detection and automated reporting to surface member-vs-casual usage patterns, seasonal trends, and station rebalancing needs."*
