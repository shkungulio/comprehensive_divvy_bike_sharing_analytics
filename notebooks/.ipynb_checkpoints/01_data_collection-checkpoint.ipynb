# 01 — Data Collection


This notebook implements the **Data Collection** phase of the CRISP-DM workflow for the
Divvy bike-share project. Its only job is to turn Divvy's public, unauthenticated S3 bucket of
monthly trip-history exports into a single local folder of consolidated CSV files that
`02_database_design.ipynb` and `03_etl_pipeline.ipynb` can build on.

The pipeline:

1. **Lists** every object currently published in the `divvy-tripdata` S3 bucket by querying its
   public XML index (no credentials required).
2. **Filters** that list down to only the monthly ZIPs between a configurable `START_YEAR`/
   `START_MONTH` and the current month, so re-running the notebook later only ever adds new
   months instead of re-downloading everything.
3. **Downloads** each remaining ZIP to disk, streaming it in chunks so large files don't have to
   fit in memory, and skipping any ZIP that's already been downloaded.
4. **Extracts** each ZIP into its own subfolder, skipping ZIPs that have already been extracted.
5. **Consolidates** every extracted `*.csv` into one flat `csv_master` folder, ready to be picked
   up by the next notebook in the pipeline.

Steps 3–5 are all **idempotent** (they check for existing files/folders before doing work), so
this notebook is safe to re-run every month as Divvy publishes new data without redoing or
duplicating prior downloads.


### Import Libraries

Only standard library modules plus three small third-party helpers: `requests` for HTTP calls,
`bs4` (BeautifulSoup) to parse the XML bucket listing, and `tqdm` for progress bars during the
download/extract loops. No credentials or API keys are needed — the Divvy S3 bucket is public.


```python
# Standard library modules
import os                      # Work with file paths and directories
import requests                # Send HTTP requests to download data from the web
from bs4 import BeautifulSoup  # Parse XML/HTML responses from the Divvy S3 bucket
from datetime import datetime  # Get the current date to limit downloads to available months
from tqdm import tqdm          # Display progress bars for long-running loops
import zipfile                 # Handle ZIP file extraction
import shutil                  # Copy files between folders
import xml.etree.ElementTree as ET  # XML utilities (not directly used but available if needed)
```

### Configuration

Defines every constant the rest of the notebook depends on: the S3 bucket's list endpoint, the
local folder layout (`downloads/` for raw ZIPs, `extracted/` for their unzipped contents,
`csv_master/` for the final consolidated CSVs), and the earliest year/month of data to pull.
`BASE_FOLDER` points at a Google Drive path — update it to match wherever you want the ~1–2 GB of
monthly exports stored locally. `START_YEAR`/`START_MONTH` let you re-scope how far back the
collection goes without touching any other cell.


```python
# Base URL for the Divvy S3 bucket listing (XML index of all available files)
BASE_URL = "https://divvy-tripdata.s3.amazonaws.com/?list-type=2"

# Root folder on local machine / Google Drive where all data will be stored
BASE_FOLDER = r"G:\My Drive\divvy_tripdata"

# Folder to store downloaded ZIP files
DOWNLOAD_FOLDER = os.path.join(
    BASE_FOLDER,
    "downloads"
)

# Folder to store extracted contents from the ZIP files
EXTRACT_FOLDER = os.path.join(
    BASE_FOLDER,
    "extracted"
)

# Folder to consolidate all CSV files from all months
CSV_FOLDER = os.path.join(
    BASE_FOLDER,
    "csv_master"
)

# Starting year and month for data to download
# Only files from this year/month onward will be considered
START_YEAR = 2024
START_MONTH = 1
```

### Create Local Directories

Creates the three folders defined above (`DOWNLOAD_FOLDER`, `EXTRACT_FOLDER`, `CSV_FOLDER`) if
they don't already exist. `exist_ok=True` makes this safe to re-run — it's a no-op on every run
after the first.


```python
# Ensure that all required folders exist before downloading or extracting
# If a folder already exists, os.makedirs with exist_ok=True will not raise an error
for folder in [
    DOWNLOAD_FOLDER,
    EXTRACT_FOLDER,
    CSV_FOLDER
]:
    os.makedirs(folder, exist_ok=True)
```

## Step 1 — Discover Available Files on S3

Queries the bucket's `?list-type=2` endpoint, which returns an XML document listing every object
("key") stored in the bucket — one key per monthly ZIP (e.g. `202401-divvy-tripdata.zip`).
BeautifulSoup parses that XML and pulls out the text of every `<Key>` tag into a plain Python
list, giving a full inventory of what's available to download before any filtering happens.


```python
# Connect to the Divvy S3 bucket and retrieve the XML listing of all objects
print("Connecting to Divvy S3 bucket...")

# Request the XML index of available files
response = requests.get(BASE_URL)

# Parse the XML response to extract file names (keys)
soup = BeautifulSoup(response.text, "xml")

# Build a list of all file keys (e.g., '202401-divvy-tripdata.zip', etc.)
files = [
    key.text
    for key in soup.find_all("Key")
]

# Display how many ZIP files were found in the bucket
print(f"Found {len(files)} ZIP files")
```

## Step 2 — Filter Files by Date Range

Divvy's filenames start with a `YYYYMM` prefix, so the year/month are parsed straight out of the
first six characters of each key. A file is kept only if it falls on or after `START_YEAR`/
`START_MONTH` **and** on or before the current year/month (guarding against any partial or
future-dated files appearing in the listing). Filenames that don't match the expected pattern are
silently skipped via the `try`/`except`. The surviving list is sorted chronologically so later
steps download/extract/report on the data in a predictable, month-by-month order.


```python
# Get the current year and month to avoid requesting future data
current_date = datetime.now()
current_year = current_date.year
current_month = current_date.month

# List to hold the subset of files that fall within the desired date range
target_files = []

# Loop through each file key and select only those between START_YEAR/MONTH and current date
for file in files:
    try:
        # Extract year and month from the file name prefix (assumes 'YYYYMM...' format)
        date_part = file[:6]
        year = int(date_part[:4])
        month = int(date_part[4:6])

        # Include files from the start date forward
        if (
            (year > START_YEAR)
            or
            (year == START_YEAR and month >= START_MONTH)
        ):
            # Exclude files beyond the current year/month (future data)
            if (
                year < current_year
                or
                (year == current_year and month <= current_month)
            ):
                target_files.append(file)

    except:
        # If the file name does not match the expected date pattern, ignore it
        pass

# Sort the selected files to download them in chronological order
target_files.sort()

# Print how many files will actually be downloaded
print(
    f"Files to download: {len(target_files)}"
)
```

## Step 3 — Download ZIP Files

For every file in `target_files`, builds the full S3 URL and the matching local path, then skips
it if that path already exists — so previously-downloaded months are never re-fetched. Otherwise
it streams the response (`stream=True`) and writes it to disk in 1 MB chunks rather than loading
the whole ZIP into memory at once, which keeps memory use flat regardless of file size.
`r.raise_for_status()` ensures a failed HTTP request (e.g. a typo'd URL or a bucket outage) raises
immediately instead of silently saving an error page as if it were a ZIP. `tqdm` wraps the loop to
show download progress across all files.


```python
# Iterate through filtered file list and download each ZIP file if not already present
for filename in tqdm(
    target_files,
    desc="Downloading files"
):
    # Build the full download URL for the current file
    url = (
        "https://divvy-tripdata.s3.amazonaws.com/"
        + filename
    )

    # Local path where the ZIP file will be stored
    zip_path = os.path.join(
        DOWNLOAD_FOLDER,
        filename
    )

    # Skip downloading if the file already exists locally
    if os.path.exists(zip_path):
        # Continue to the next file without re-downloading
        continue

    # Stream the file from the server to avoid loading it fully into memory
    with requests.get(
        url,
        stream=True
    ) as r:

        # Raise an exception if the request was not successful
        r.raise_for_status()

        # Get the total size from the headers (if available)
        total_size = int(
            r.headers.get(
                "content-length",
                0
            )
        )

        # Write the response content to disk in chunks
        with open(
            zip_path,
            "wb"
        ) as f:

            for chunk in r.iter_content(
                chunk_size=1024*1024
            ):
                # Write each chunk as it arrives
                f.write(chunk)
```

## Step 4 — Extract ZIP Files

Unzips each downloaded file into its own subfolder under `EXTRACT_FOLDER`, named after the ZIP
(e.g. `202401-divvy-tripdata/`) so months never collide or overwrite each other. As with the
download step, a ZIP is skipped if its extraction folder already exists, making this step
idempotent on repeat runs.


```python
# Extract each downloaded ZIP file into its own subfolder
print("Extracting files...")

for filename in tqdm(
    target_files,
    desc="Extracting"
):

    # Full path to the downloaded ZIP file
    zip_path = os.path.join(
        DOWNLOAD_FOLDER,
        filename
    )

    # Folder where the contents of this ZIP will be extracted
    extract_path = os.path.join(
        EXTRACT_FOLDER,
        filename.replace(".zip","")
    )

    # Skip extraction if this ZIP has already been extracted
    if os.path.exists(extract_path):

        continue

    # Open the ZIP file and extract all contents into the target folder
    with zipfile.ZipFile(
        zip_path,
        "r"
    ) as zip_ref:

        zip_ref.extractall(
            extract_path
        )
```

## Step 5 — Consolidate CSV Files

Walks the entire `EXTRACT_FOLDER` tree (each month's own subfolder) and copies every real `.csv`
file into one flat `CSV_FOLDER`, which is the input `02_database_design.ipynb` and
`03_etl_pipeline.ipynb` expect. macOS resource-fork artifacts (files starting with
`._`) are explicitly skipped since some ZIP tools include them alongside the real CSV and they
would otherwise pollute the master folder. `shutil.copy2` preserves file metadata (e.g. modified
time) when copying.


```python
# Walk through the extracted folders and copy all CSV files into a single master folder
print("Collecting CSV files...")

for root, dirs, files in os.walk(EXTRACT_FOLDER):

    for file in files:

        # Skip macOS metadata files
        if file.startswith("._"):
            continue

        # Only copy real CSV files
        if file.endswith(".csv"):

            source = os.path.join(root, file)
            destination = os.path.join(CSV_FOLDER, file)

            shutil.copy2(source, destination)

print("\nCOMPLETE!")
print(f"CSV files stored in: {CSV_FOLDER}")
```
